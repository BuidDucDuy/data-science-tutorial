pyplot api và object-oriented (oo) api

```python
# pyplot api 
x = np.linspace(0 , 5 , 1000 , dtype = None)
x
```

```python
plt.plot(x , np.sin(x) , color = "red" , label = "sinx") ; 
plt.plot(x , np.cos(x) , color = "blue" ,label = "cosx");
plt.xlabel("Trục x")
plt.ylabel("Trục y")
# plt.ylim([0,4])
# plt.xlim([0,4])
plt.legend() ;

```

object-oriented (oo) api

```python
# prepare data 
x = [11,22,39,40]
y = [1,2,3,4]

# setup hình tổng
fig , ax = plt.subplots(figsize = (10,6))
ax.plot(x,y)
```
