```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cities = pd.read_csv("../data/california_cities.csv")
rename_dict = {
    "Unnamed: 0": "STT",
    "city": "Thành phố",
    "latd": "Vĩ độ",
    "longd": "Kinh độ",
    "elevation_m": "Độ cao (m)",
    "elevation_ft": "Độ cao (ft)",
    "population_total": "Dân số",
    "area_total_sq_mi": "Diện tích tổng (mi²)",
    "area_land_sq_mi": "Diện tích đất (mi²)",
    "area_water_sq_mi": "Diện tích nước (mi²)",
    "area_total_km2": "Diện tích tổng (km²)",
    "area_land_km2": "Diện tích đất (km²)",
    "area_water_km2": "Diện tích nước (km²)",
    "area_water_percent": "Tỷ lệ nước (%)"
}
cities.rename(columns=rename_dict , inplace=True)
cities.head()

```

```python
#xuất kinh độ và vĩ độ
kinh , vi = cities["Kinh độ"] , cities["Vĩ độ"] ;
fig ,ax = plt.subplots()
plt.style.use('seaborn-v0_8-whitegrid')
ax.scatter(kinh , vi , s = 20)
plt.xlabel("Kinh độ")
plt.ylabel("Vĩ độ")
```

```python
#dân số và diện tích
kinh , vi = cities["Kinh độ"] , cities["Vĩ độ"] ;
danso , dientich = cities["Dân số"] , cities["Diện tích tổng (km²)"]
plt.scatter(kinh , vi , c = np.log10(danso),cmap='tab10' , s = dientich,label="Thành phố");
plt.axis("equal")
plt.xlabel("Kinh độ")
plt.ylabel("Vĩ độ")
plt.colorbar(label = 'log$_{10}$(danso)')
plt.legend() ;

```
