```python
import numpy as np
```

```python
np.array([1,2,3,4,5,6,7])
```

```python
np.zeros([2,3],dtype = int , order="f")
```

```python
np.ones([2,3],dtype=int,order="c")
```

```python
a1 = np.array([[1,2,3],
              [4,5,6]])
a2 = np.array([[1,2,3],
              [4,5,6]])
print("a1:" , a1.flatten(order="c"))
print("a2:" , a2.flatten(order="f"))
```

```python
np.arange(0,20,2)
```

```python
np.full([3,5] , 7)
```

```python
np.linspace(0,2,5,dtype = float)
```
