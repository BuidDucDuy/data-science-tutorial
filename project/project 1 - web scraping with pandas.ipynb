```python
# rename
df_premier24.head()
df_premier24.rename(
    columns = {
        'Rk':"Rank" ,
        'Player':"tên cầu thủ" , 
        'Nation':'Quốc tịch',
        "Squad": "Câu_lạc_bộ",
        "Age": "Tuổi",
        "Born": "Năm_sinh",
        "MP": "Số_trận",
        "Starts": "Đá_chính",
        "Min": "Phút_thi_đấu",
        "90s": "Số_trận_90p",
        "Gls.1": "Bàn_thắng",
        "Ast.1": "Kiến_tạo",
        "G+A.1": "Bàn+KT",
        "G-PK.1": "Bàn_không_pen",
        "PK": "Pen_thành_công",
        "PKatt": "Sút_pen",
        "CrdY": "Thẻ_vàng",
        "CrdR": "Thẻ_đỏ",
        "xG.1": "Bàn_kỳ_vọng",
        "npxG.1": "Bàn_không_pen_kỳ_vọng",
        "xAG.1": "Kiến_tạo_kỳ_vọng",
        "npxG+xAG.1": "Hiệu_suất_tấn_công_kỳ_vọng",
        "PrgC": "Dẫn_bóng_tiến",
        "PrgP": "Chuyền_tiến",
        "PrgR": "Nhận_bóng_tiến"
    },
  inplace = True 
)
```

```python
df_premier24
```

```python
# hiện 1 só cột
df_premier24 = df_premier24[['Rank' , 'tên cầu thủ','Rank' , 'Quốc tịch','Pos' , 'Tuổi','Năm_sinh' , 'Số_trận','Đá_chính' , 'Phút_thi_đấu' , 'Bàn_thắng']]
```

```python
df_premier24.head()
```

```python
#hiện bàn thắng trung bình
ban_thang_TB_player = (df_premier24.groupby('tên cầu thủ')['Bàn_thắng'].transform('mean'))
df_premier24['Bàn_thắng_TB_player'] = ban_thang_TB_player
df_premier24 = df_premier24.drop(columns=['Bàn_thắng_TB'])
df_premier24.head(15)
```
